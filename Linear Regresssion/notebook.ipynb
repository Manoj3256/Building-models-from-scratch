```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#data:-
samples=100
data_X=np.random.rand(samples)*10
data_y=2*data_X+1+np.random.randn(samples)*2
```

```python
plt.plot(data_X,data_y,'r.',label='Train data')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Train data')
plt.legend()
plt.show()
```

```python
#Model
model=LinearRegression()
model.fit(data_X.reshape(-1,1),data_y)
pre=model.predict(data_X.reshape(-1,1))
model.intercept_,model.coef_
```

```python
plt.plot(data_X,data_y,'b.',label='Trained data')
plt.plot(data_X,pre,'r-',label='Predicted data')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Result')
plt.legend()
plt.savefig("Images/Predicted_data.png")
plt.show()
```
