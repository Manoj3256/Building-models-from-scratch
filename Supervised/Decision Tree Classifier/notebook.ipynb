```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
from sklearn.metrics import recall_score,precision_score,f1_score
data=load_iris(as_frame=True)
```

```python
data_train_X,data_test_X,data_train_y,data_test_y=train_test_split(data.data,
    data.target,shuffle=True,train_size=.8)
```

```python
data_train_y.head()
```

```python
model=DecisionTreeClassifier(max_depth=3)
model.fit(data_train_X,data_train_y)
pred=model.predict(data_test_X)
```

```python
precision_score(data_test_y, pred, average='macro')
```

```python
recall_score(data_test_y, pred, average='macro')
```

```python
f1_score(data_test_y, pred, average='macro')
```
