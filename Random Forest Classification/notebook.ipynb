```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import recall_score,precision_score
data=load_iris(as_frame=True)
data_X,data_y=data.data,data.target
```

```python
data_X.info()
```

```python
train_X,test_X,train_y,test_y=train_test_split(data_X,data_y,train_size=.75,random_state=42)
len(train_X),len(test_X)
```

```python
model=RandomForestClassifier(n_estimators=50,random_state=42)
model.fit(train_X,train_y)
pred=model.predict(test_X)
```

```python
precision_score(pred,test_y,average='weighted')
```

```python
recall_score(pred,test_y,average="weighted") #Since the dataset is small, the model is overfitting(memorizing) the data
```

```python
for sc,name in zip(model.feature_importances_,data.data.columns):
    print(f"{name}:{round(sc,2)}")
```
